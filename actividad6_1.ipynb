```python
import os
if not os.path.exists("archivo.txt"):
    open("archivo.txt", "w").close()
import tkinter as tk
from tkinter import ttk
from tkinter import messagebox
root = tk.Tk()
root.geometry("300x200")
header = tk.Frame(root)
header.pack(side="top", fill="x")
empleados=[]
def agregar():
    win = tk.Toplevel(root)
    win.geometry("500x500")
    nombre = tk.Entry(win, bg="#D6F7FF", width=40)
    nombre.insert(0, "Nombre")
    nombre.place(x=150, y=0)
    apellido = tk.Entry(win, bg="#D6F7FF", width=40)
    apellido.insert(0, "Apellido")
    apellido.place(x=150, y=20)
    sueldo = tk.Entry(win, bg="#D6F7FF", width=40)
    sueldo.insert(0, "Pago/dia")
    sueldo.place(x=150, y=40)
    dias = tk.Spinbox(win, from_=1, to=31)
    dias.place(x=150, y=60)

    genero = tk.StringVar()
    rb_m = tk.Radiobutton(win, text="Masculino", variable=genero, value="M")
    rb_f = tk.Radiobutton(win, text="Femenino", variable=genero, value="F")
    rb_m.place(x=150, y=80)
    rb_f.place(x=240, y=80)

    opciones = ["Directivo", "Estratégico", "Operativo"]
    combo = ttk.Combobox(win, values=opciones, state="readonly")
    combo.current(0)
    combo.place(x=150, y=110)

    Otros = tk.Entry(win, bg="#D6F7FF", width=40)
    Otros.insert(0, "Otros ingresos")
    Otros.place(x=150, y=130)

    Pagos_salud = tk.Entry(win, bg="#D6F7FF", width=40)
    Pagos_salud.insert(0, "Pagos por salud")
    Pagos_salud.place(x=150, y=150)

    Aporte = tk.Entry(win, bg="#D6F7FF", width=40)
    Aporte.insert(0, "Aporte pensiones")
    Aporte.place(x=150, y=170)

    boton1 = tk.Button(win, text="Enviar", command=lambda: enviar_1(
            nombre.get(),
            apellido.get(),
            genero.get(),
            combo.get(),
            sueldo.get(),
            dias.get(),
            Otros.get(),
            Pagos_salud.get(),
            Aporte.get()), bg="#D6F7FF", width=10)
    boton1.place(x=150, y=190)
def enviar_1(nombre, apellido, genero, combo, sueldo, dias, otros, salud, aporte):
    mensaje=""
    error=0
    try:
        if len(nombre) == 0 or len(apellido) == 0 or len(sueldo) == 0 or len(dias) == 0 \
            or len(genero) == 0 or len(combo) == 0 or len(otros) == 0 \
            or len(salud) == 0 or len(aporte) == 0:
            raise ValueError("Por favor rellene todas las celdas")
    except ValueError as e:
        messagebox.showerror("Fallo", e)

    try:
        float(sueldo)
    except ValueError:
        error=1
        mensaje=mensaje+"/El suedo debe ser numerico/  "
    try:
        float(otros)
    except ValueError:
        error=1
        mensaje=mensaje+"/Otros ingresos debe ser numerico/  "
    try:
        float(salud)
    except ValueError:
        error=1
        mensaje=mensaje+"/La salud debe ser numerica/  "
    try:
        float(aporte)
    except ValueError:
        error=1
        mensaje=mensaje+"/El aporte pensiones debe ser numerico/  "
    if error==1:
        messagebox.showerror("Error", f'{mensaje}')
    else:
        messagebox.showinfo("Resultados", f'Usuario creado')
        vec.append(usuario_nuevo(nombre, apellido, dias, sueldo, otros, salud, aporte))
class usuario_nuevo:
    def __init__(self,nombre, apellido, dias, sueldo, otros, salud, aporte):
        self.nombre=nombre
        self.apellido=apellido
        sueldo_f= float(dias)*float(sueldo)+float(otros)-float(salud)-float(aporte)
        self.salario=sueldo_f


def Nomina():
    win = tk.Toplevel(root)
    win.title("Nómina")
    win.geometry("400x400")

    texto = tk.Text(win, width=50, height=20, bg="#F0FFFF")
    texto.pack(pady=10)

    # Mostrar archivo.txt
    with open("archivo.txt", "r") as file:
        lineas = file.readlines()
        if len(lineas) == 0:
            texto.insert("end", "No hay empleados guardados.\n")
        else:
            for linea in lineas:
                datos = linea.strip().split(",")
                texto.insert("end", f"{datos[0]} {datos[1]} {datos[2]}\n")

    texto.insert("end", "")

    # Mostrar vec
    if len(vec) == 0:
        texto.insert("end", "a")
    else:
        for p in vec:
            texto.insert("end", f"{p.nombre} {p.apellido} {p.salario}\n")



def Guardar():
    with open("archivo.txt", "a") as file:
        for p in vec:
            file.write(f"{p.nombre},{p.apellido},{p.salario}\n")
    messagebox.showinfo("Completado", "Personas guardadas")
    vec.clear()

vec=[]
boton1 = tk.Button(header, text="Crear", command=agregar, bg="#D6F7FF", width=10)
boton1.pack(side="left")
boton1 = tk.Button(header, text="Nomina", command=Nomina, bg="#D6F7FF", width=10)
boton1.pack(side="left")
boton1 = tk.Button(header, text="Guardar", command=Guardar, bg="#D6F7FF", width=10)
boton1.pack(side="left")
root.mainloop()
```
